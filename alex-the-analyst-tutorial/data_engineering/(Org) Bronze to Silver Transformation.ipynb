```python
import pandas as pd

# load the data from the table
df = spark.table('data_engineering.video1.users_dirty_csv').toPandas()
```

```python
# Fix the date column, replace '.' with '/' separator
df['signup_date'] = df['signup_date'].str.replace('.','/',regex=False)

# covert to datetime format
df['signup_date'] = pd.to_datetime(df['signup_date'], format='%m/%d/%y', errors='coerce')

# remove duplicates based on user_id, keeping the first occurence
df_clean = df.drop_duplicates(subset = ['user_id'], keep='first')
```

```python
display(df_clean)
```

```python
# create pandas dataframe back to spark dataframe
df_clean_spark = spark.createDataFrame(df_clean)

# write to silver table in the same schema
df_clean_spark.write.mode('overwrite').saveAsTable('data_engineering.video1.users_cleaned')
```
